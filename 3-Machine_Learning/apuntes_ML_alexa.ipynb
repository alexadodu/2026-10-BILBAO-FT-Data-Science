## 1.1 Machine Learning supervisado

Tipo de machine learning en el que el modelo **se entrena con un conjunto de datos etiquetados** (es decir, **se conoce la variable objetivo o de resultado**). 



> Con un modelo puedo predecir cierto output, como por ejemplo si el paciente contrae o no cáncer de pulmón, a partir de un conjunto de inputs, que podrían ser edad, fumador, enfermedades previas...

### Glosario

- **Target:** Se trata de la variable objetivo, la variable a predecir, el output del modelo. Por norma general es una única variable, ya sea numérica o categórica.
- **Feature:** Todo el resto de variables utilizadas como input del modelo.

### División del dataset en Train-Test

Al crear modelos de aprendizaje automático, estos se entrenan con un conjunto de datos específico (X e y). Una vez entrenados, es importante asegurarse de que el modelo también funcione correctamente con datos de prueba desconocidos. **La división de los datos en conjuntos de entrenamiento y prueba permite comprobar si el modelo de aprendizaje automático funciona bien con datos que no ha visto**.



Entonces tienes un conjunto de datos que contiene las etiquetas (y) y los predictores (características X). Divide el conjunto de datos aleatoriamente en dos subconjuntos:

- Conjunto de entrenamiento: Entrenar el modelo de aprendizaje automático
- Conjunto de pruebas: Compruebe la precisión del modelo.

En el primer subconjunto, denominado conjunto de entrenamiento, entrenarás el algoritmo de aprendizaje automático y construirás el modelo de aprendizaje automático.

Luego, utilice este modelo de aprendizaje automático en el otro subconjunto, denominado conjunto de prueba, para predecir las etiquetas. A continuación, podrá comparar las etiquetas predichas con las etiquetas reales del conjunto de prueba.

Si el modelo es bueno, las predicciones de etiquetas se aproximarán a las etiquetas reales.
