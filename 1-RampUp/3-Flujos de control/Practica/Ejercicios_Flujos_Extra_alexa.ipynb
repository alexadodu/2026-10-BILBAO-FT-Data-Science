## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
sigue = True
resultados = [] # aquí se recogen los resultados de cada operación (para el historial)

# mientras el buclea sea True, se va a ejecutar la calculadora
while sigue:
    # guarda la operación matemática que se quiere realizar y además, tiene en cuenta si escribe con tildes
    # y mayúsculas
    operacion = input("¿Qué operación quieres realizar? (multiplicación/división/suma/resta/potencia/salir)"
                      ).lower().replace("ó", "o")
    operaciones = ["multiplicacion", "division", "suma", "resta", "potencia"]

    # si la operación no es salir, es decir, está en la variable operaciones, va a pedirle 
    # dos números al usuario y va a intentar pasar el número a float. Si tiene letras o comas,
    # le dice que el número no es válido.
    if operacion in operaciones:
        try:
            num1 = float(input("Ingresa un número: "))
            num2 = float(input("Ingresa otro número: "))
        except ValueError:
            print("Introduce un nº válido, sin letras, comas, espacios en blanco ni otros.")
            continue
    
    # en este caso, si la operación es salir, se para el bucle (con sigue = False) y 
    # va a pedirle al usuario si quiere imprimir por pantalla el historial de todas
    # las operaciones realizadas
    elif operacion == "salir":
        sigue = False
        imprimir = input("¿Quieres imprimir por pantalla todas las operaciones realizadas? (sí/no)"
                         ).lower().replace("í", "i")
        if imprimir == "si":
            print("--- HISTORIAL ---")
            print(*resultados, sep="\n")
    
    # si la operación no coincide con las operaciones que se le permite, le dice al usuario que no se puede
    else:
        print("Esa operación no es posible.")

    # según cada operación, se va a imprimir por pantalla la operación realizada con su resultado y, además,
    # se guarda la operación en la lista 'resultados' (historial)
    match operacion:
        case "multiplicacion":
            resultado = num1 * num2
            print_resultado = f"{num1} x {num2} = {resultado}"
            resultados.append(print_resultado)
            print(print_resultado)
                    
        case "division":
            # como no se puede dividir por 0, se recoge ese error para avisar del ello al usuario
            try:
                resultado = num1 / num2
                print_resultado = f"{num1} / {num2} = {resultado}"
                resultados.append(print_resultado)
                print(print_resultado)
            except Exception as e:
                print("No se puede dividir por 0.")
        
        case "suma":
            resultado = num1 + num2
            print_resultado = f"{num1} + {num2} = {resultado}"
            resultados.append(print_resultado)
            print(print_resultado)
        
        case "resta":
            resultado = num1 - num2
            print_resultado = f"{num1} - {num2} = {resultado}"
            resultados.append(print_resultado)
            print(print_resultado)
        
        case "potencia":
            resultado = num1 ** num2
            print_resultado = f"{num1}^{num2} = {resultado}"
            resultados.append(print_resultado)
            print(print_resultado)

```
