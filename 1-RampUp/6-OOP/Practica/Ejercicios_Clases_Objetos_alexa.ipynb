## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomésticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas últimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un número con todas las ventas.
* Método que calcula la media de ventas de los últimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
class Tienda:
    # La clase tienda tiende dos atributos comunes:
    tipo = "electrodomésticos"
    abierta = True

    # Aquí se definen los atributos único de cada tienda:
    def __init__(self, nombre, direccion, num_empleados, ventas_ultimos_3_meses = [1, 1, 1]):
        self.nombre = str(nombre)
        self.direccion = str(direccion)
        self.num_empleados = int(num_empleados)
        self.ventas = ventas_ultimos_3_meses

    # Desde aquí empiezan los métodos para calcular diferentes cosas
    def calculo_ventas(self):
        return sum(self.ventas)

    def calculo_ventas_media_empleado(self):
        media = sum(self.ventas) / len(self.ventas)
        return round((media / self.num_empleados), 2)

    def nombre_direccion(self):
        return f"Nombre de la tienda: {self.nombre} - Dirección: {self.direccion}"

    def ventas_ultimo_mes(self):
        return self.ventas[-1]

    def proyeccion_ventas(self, x):
        if x < 1000:
            for indice, ventas_mes in enumerate(self.ventas):
                ventas_mes *= 1.2
                self.ventas[indice] = round(ventas_mes, 2)
            return self.ventas
        elif x >= 1000:
            for indice, ventas_mes in enumerate(self.ventas):
                ventas_mes *= 1.5
                self.ventas[indice] = round(ventas_mes, 2)
            return self.ventas
pass
```

```python
pepco = Tienda("pepCO", "Avenida", 20, [10000.2, 30008.32, 20090.54])
mediaMarkt = Tienda("Media Markt", "Donosti", 2000, [343290.889, 239890.23, 3948124.42])
barrio = Tienda("La de tu barrio", "Zazpi kaleak", 3, [1300.83, 3043.67, 2577.68])

print("--- Probando todo lo implementado en la clase 'Tienda' con una tienda ---")
print("*** Atributos:")
print(pepco.tipo)
print(pepco.abierta)
print(pepco.nombre)
print(pepco.direccion)
print(pepco.num_empleados)
print(pepco.ventas)
print("*** Métodos:")
print(pepco.calculo_ventas())
print(pepco.calculo_ventas_media_empleado())
print(pepco.nombre_direccion())
print(pepco.ventas_ultimo_mes())
print(pepco.proyeccion_ventas(1000))

print("\n--- Cálculo de las ventas del último mes de todas las tiendas ---")
tiendas = [pepco, mediaMarkt, barrio]
for tienda in tiendas:
    print(f"Ventas del último mes de {tienda.nombre}: {tienda.ventas_ultimo_mes()}€.")

print("\n--- Imprimiendo por pantalla los nombres de las tiendas cuya dirección lleve el string 'Avenida' ---")
for tienda in tiendas:
    if tienda.direccion == "Avenida":
        print(tienda.nombre)
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
    # Atributos que sabemos seguro que tiene un perro
    patas = 4
    orejas = 2
    ojos = 2
    velocidad = 0

    # Atributos únicos de cada perro
    def __init__(self, raza, pelo = "marrón", duenio = False):
        self.raza = raza
        self.pelo = pelo
        self.duenio = duenio

    # método para andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
    def andar(self, aumento_velocidad):
        return self.velocidad + aumento_velocidad

    def parar(self):
        self.velocidad = 0
        return self.velocidad

    # método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.
    def ladrar(self, palabra):
        return "GUAU! " + palabra
pass
```

```python
perro = Perro("labrador")

print(perro.raza)
print(perro.pelo)
print(perro.duenio)
print(perro.patas)
print(perro.orejas)
print(perro.ojos)
print(perro.velocidad)
print()

print(perro.andar(2))
print(perro.parar())
print(perro.ladrar("Paco"))
```
