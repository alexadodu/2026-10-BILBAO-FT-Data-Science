## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def que_dia_es(arg):
    match arg:
        case 1:
            task = "Lunes"
        case 2:
            task = "Martes"
        case 3:
            task = "Miércoles"
        case 4:
            task = "Jueves"
        case 5:
            task = "Viernes"
        case 6:
            task = "Sábado"
        case 7:
            task = "Domingo"
        case _:
            task = "no es un día válido"
    return task
````

````python
que_dia_es(3)
````

## Ejercicio 3
Escribe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero sea  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def comparacion_dos_num(num1, num2):
    if num1 == num2:
        return "son iguales"
    if num1 > num2:
        return "el primero es mayor que el segundo"
    else:
        return "el segundo es mayor que el primero"
````

````python
comparacion_dos_num(10, 10)
````

````python
comparacion_dos_num(10, 1)
````

````python
comparacion_dos_num(1, 10)
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python
def contador_letra(texto, letra):
    texto = texto.lower()
    letra = letra.lower()
    return int(texto.count(letra))
````

````python
contador_letra("probemos a Contar la letra A", "A")
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string, siendo la clave la letra y el valor el conteo.

````python
def conteo_letras_dic(string):
    dic = {}
    for c in string.lower():
        dic[c] = string.count(c)
        
    return dic
````

````python
conteo_letras_dic("Estamos buscando un/a Investigador/a de Inteligencia Artificial altamente motivado/a para unirse a nuestro equipo de tecnología.")
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

Tendrá como salida la lista

````python
def modif_lista(lista, comando, elemento = None):
    if comando == "add" and elemento:
        lista.append(elemento)
        return lista
    elif comando == "remove" and elemento:
        lista.remove(elemento)
        return lista
    else:
        return lista
    
````

````python
modif_lista(lista = [1,2,3,4], comando = "add")
````

````python
modif_lista(lista = [1,2,3,4], comando = "add", elemento = 5)
````

````python
modif_lista(lista = [1,2,3,4], comando = "remove", elemento = 3)
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)

````python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return (fibonacci(n-1) + fibonacci(n-2))
````

````python
fibonacci(10)
````

## Ejercicio 9
Define en una única celda las siguientes funciones:
* Función que calcule el área de un cuadrado
* Función que calcule el área de un triángulo
* Función que calcule el área de un círculo

En otra celda, calcular el area de:
* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python
import math
````

````python
def area_cuadrado(lado):
    return lado**2

def area_triangulo(base, altura):
    return (base * altura) / 2

def area_circulo(radio):
    return math.pi * (radio **2)
````

````python
print(area_circulo(10) + area_triangulo(3, 7))

print(area_cuadrado(10) + area_circulo(4) + (area_circulo(6) * 2) + (5 * area_triangulo(2,4)))
````
