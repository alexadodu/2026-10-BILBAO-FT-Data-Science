## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
import random

palabras_naturaleza = [
    "bosque", "montana", "rio", "oceano", "valle", 
    "desierto", "selva", "cascada", "volcan", "llanura", 
    "isla", "acantilado", "canon", "glaciar", "pradera", 
    "arbol", "flor", "planta", "musgo", "raiz", 
    "hoja", "semilla", "hierba", "arbusto", "helecho", 
    "brote", "corteza", "animal", "pajaro", "insecto", 
    "pez", "mamifero", "reptil", "especie", "nido", 
    "garra", "pluma", "lluvia", "viento", "tormenta", 
    "nieve", "trueno", "niebla", "rayo", "relampago", 
    "tierra", "cielo", "piedra", "ecosistema", "biodiversidad"
    ]

def ahorcado():
    """
    Ejecuta el juego del ahorcado.
    
    Selecciona una palabra aleatoria sobre la naturaleza, gestiona los intentos
    del usuario validando las entradas, controla el sistema de vidas y muestra 
    visualmente el progreso de la palabra oculta hasta que se gana o se pierde.
    """
    # Selecciona una palabra al azar de la lista y la convierte a mayúsculas
    palabra_aleatoria = random.choice(palabras_naturaleza).upper()
    # print(palabra_aleatoria)
    
    # Muestra inicialmente la longitud de la palabra con guiones bajos
    print("_ " * len(palabra_aleatoria))
    print()

    # Inicialización de las variables de control del juego
    vidas = 6
    letras_input = set()      # Historial de todas las letras ingresadas por el usuario
    ganar = False

    # Bucle principal del juego: se ejecuta mientras el jugador tenga vidas
    while vidas > 0:

        # Bucle de validación: asegura que el usuario ingrese una letra válida y nueva
        while True:
            letra_usuario = input("Inserta una letra (sin tildes): ").upper()    

            # Comprueba que la entrada sea una sola letra y no un número o cadena larga
            if letra_usuario.isalpha() and len(letra_usuario) == 1:
                # Verifica si el usuario ya había intentado esa letra antes
                if letra_usuario in letras_input:
                    print(f"Ya has ingresado {letra_usuario}. Prueba nuevamente.")
                else:
                    # Registra la nueva letra en el historial y rompe el bucle de entrada
                    letras_input.add(letra_usuario)
                    break
            else:
                # Informa al usuario si la entrada no cumple los requisitos de formato
                print(f"\nPor favor, ingresa nuevamente una letra. {letra_usuario} es incorrecto.")

        print(f"\nHistorial de letras ingresadas: {letras_input}.")
                
        # Evaluación del intento: comprueba si la letra propuesta está en la palabra secreta
        if letra_usuario in palabra_aleatoria:
            ganar = True  # Hipótesis inicial: asumimos que ha completado la palabra

            # Dibuja el estado actual de la palabra en pantalla
            for c in palabra_aleatoria:
                if c == letra_usuario:
                    print(letra_usuario, end=" ")
                elif c in letras_input:
                    print(c, end= " ")
                else:
                    print("_", end=" ")
                    ganar = False  # Si queda algún guion bajo, aún no ha ganado por completo
            print()
        else:
            # Si la letra es incorrecta, penaliza restando una vida
            vidas -= 1
            print(f"\nLa letra {letra_usuario} no está, Te quedan {vidas} vidas.")

        # Condición de victoria: si la bandera 'ganar' sigue en True, termina el juego
        if ganar:
            print("\n¡FELICIDADES! Has adivinado la palabra.")
            break

    # Se ejecuta únicamente si el bucle 'while vidas > 0' termina de forma natural (llegar a 0 vidas)
    else:
        print(f"\nFin del juego. La palabra era {palabra_aleatoria}.")

````
