## Ejercicio 4
Realiza un conversor de grados Celsius a Fahrenheit. El conversor tiene que recibir el input del usuario y devolver la conversión.

La conversión de grados sigue la siguiente fórmula: F = C × (9 / 5) + 32 

```python
input_celsius = input("Introduce los grados Celsius a convertir en Fahrenheit: ")
farenheit = float(input_celsius) * (9 / 5) + 32
print(f"Los grados Fahrenheit correspondientes a {input_celsius}ºC son {farenheit}ºF.")
```
