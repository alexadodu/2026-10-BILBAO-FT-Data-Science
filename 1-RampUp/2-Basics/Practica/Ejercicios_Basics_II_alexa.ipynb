## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math

input_grados = input("Introduce los grados a pasar a radianes: ")
radianes = float(input_grados) * (math.pi / 180)
print(f"Los radianes correspondientes a {input_grados}° son {round(radianes, 2)} rad.")
````

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

````python
base = input("Introduce la base de tu paralelograme en cm: ")
altura = input("Introduce la altura de tu paralelograme en cm: ")
area = float(base) * float(altura)
print(f"El área de tu paraleogramo ingresado (base: {base} cm, altura: {altura} cm) es de {round(area, 2)} cm².")
````

## Ejercicio 4
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C 
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
A = 4
B = "Text"
C = 4.1

if A == B:
    print("A y B son equivalentes")
else:
    print("A y B no son equivalentes")
````

````python
if A != C:
    print("A y C no son equivalentes")
else:
    print("A y C son equivalentes")


if A > C:
    print("A > C")
else:
    print("A <= C")

if C <= A:
    print("C <= A")
else:
    print("C > A")


if B != C:
    print("B y C no son equivalentes")
else:
    print("B y C son equivalentes")
````

## Ejercicio 6
Mismo programa que en 5, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
contr1 = input("Introduce tu contraseña: ")
contr2 = input("Introduce nuevamente tu contraseña: ")
contr3 = input("Por MEGASEGURIDAD. Introduce nuevamente tu contraseña: ")

if contr1 == contr2 == contr3:
    print("Contraseña correcta. Todas iguales.")
elif contr1 == contr2 or contr1 == contr3 or contr2 == contr3:
    print("Contraseña incorrecta. Una de las contraseñas introducidas no coindice con las otras.")
else:
    print("Contraseña incorrecta. Ninguna de las contraseñas introducidas coincide con las otras.")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10, cada condición por separado

````python
num1 = input("Introduce un número: ")
num2 = input("Introduce otro número: ")

suma = float(num1) + float(num2)

if suma == 10:
    print("La suma de los dos números es igual a 10.")
elif suma < 10:
    print("La suma de los dos números es menor que 10.")
else:
    print("La suma de los dos números es mayor que 10.")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
