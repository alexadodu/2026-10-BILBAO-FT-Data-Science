Programa el SW de una biblioteca:

* Preguntar qué operación quieres realizar (visualizar todos los libros de la biblioteca/buscar libro/añadir libro/eliminar libro por título)
    * En caso de que pida el usuario elija visualizar se visualizarán todos los libros
    * Buscar libro: búsqueda por título que introduce el usuario y se visualiza todos los campos del libro
    * Añadir libro: se debe solicitar al usuario que introduzca título y autor para el nuevo libro y se añade el libro a la lista de libros
    * Eliminar libro: se solicita al usuario el título del libro y éste debe ser eliminado de la lista de libros
    * Alquilar libro: se solicita al usuario el título del libro y se debe pasar el campo "Alquilado" a True
    * Devolver libro: se solicita al usuario el título del libro y se debe pasar el campo "Alquilado" a False

```python
libros = [
    {"Titulo": "Python Data Science Handbook", "Autor": "Jake VanderPlas", "Alquilado": False},
    {"Titulo": "Hands-On Machine Learning with Scikit-Learn, Keras, and TensorFlow", "Autor": "Aurélien Géron", "Alquilado": True},
    {"Titulo": "Pattern Recognition and Machine Learning", "Autor": "Christopher M. Bishop", "Alquilado": False},
    {"Titulo": "Deep Learning", "Autor": "Ian Goodfellow, Yoshua Bengio, Aaron Courville", "Alquilado": True},
    {"Titulo": "The Elements of Statistical Learning", "Autor": "Trevor Hastie, Robert Tibshirani, Jerome Friedman", "Alquilado": False},
    {"Titulo": "Data Science for Business", "Autor": "Foster Provost, Tom Fawcett", "Alquilado": False},
    {"Titulo": "Bayesian Data Analysis", "Autor": "Andrew Gelman et al.", "Alquilado": True},
    {"Titulo": "Introduction to the Theory of Computation", "Autor": "Michael Sipser", "Alquilado": False},
    {"Titulo": "Artificial Intelligence: A Modern Approach", "Autor": "Stuart Russell, Peter Norvig", "Alquilado": True},
    {"Titulo": "Computer Vision: Algorithms and Applications", "Autor": "Richard Szeliski", "Alquilado": False},
    {"Titulo": "Data Science from Scratch", "Autor": "Joel Grus", "Alquilado": True},
    {"Titulo": "The Art of Statistics", "Autor": "David Spiegelhalter", "Alquilado": False},
    {"Titulo": "Python Machine Learning", "Autor": "Sebastian Raschka, Vahid Mirjalili", "Alquilado": True},
    {"Titulo": "An Introduction to Statistical Learning", "Autor": "Gareth James, Daniela Witten, Trevor Hastie, Robert Tibshirani", "Alquilado": False},
    {"Titulo": "Fundamentals of Data Engineering", "Autor": "Joe Reis, Matt Housley", "Alquilado": False},
    {"Titulo": "Storytelling with Data", "Autor": "Cole Nussbaumer Knaflic", "Alquilado": True},
    {"Titulo": "Building Machine Learning Powered Applications", "Autor": "Emmanuel Ameisen", "Alquilado": False},
    {"Titulo": "Practical Statistics for Data Scientists", "Autor": "Peter Bruce, Andrew Bruce", "Alquilado": True},
    {"Titulo": "SQL for Data Scientists", "Autor": "Renee M. P. Teate", "Alquilado": False},
    {"Titulo": "Data Engineering on Azure", "Autor": "Vlad Riscutia", "Alquilado": True}
]
```

```python
# ==============================================================================
# FUNCIONES DE UTILIDAD (Herramientas reutilizables)
# ==============================================================================

def busqueda_libro(lista_libros, titulo_buscar):
    """
    Busca un libro recorriendo todos los valores de cada diccionario.
    Retorna el diccionario original (por referencia) si lo encuentra, o None si no.
    """
    for libro in lista_libros:
        for valor in libro.values():
            # Convertimos a minúsculas y string para comparar sin importar el formato original
            valor_minusc = str(valor).lower()
            if titulo_buscar == valor_minusc:
                return libro  # Corta el bucle y devuelve el objeto exacto de la lista
    return None  # Si termina de revisar todo el bucle y no encuentra coincidencia


def confirmacion(pregunta_input):
    """
    Muestra una pregunta al usuario y estandariza su respuesta.
    Retorna True si responde 'si' (con o sin tilde) y False en cualquier otro caso.
    """
    # .strip() quita espacios accidentales, .replace() elimina la tilde de 'sí'
    respuesta = input(pregunta_input).lower().replace("í", "i").strip()
    return respuesta == "si"


# ==============================================================================
# BUCLE PRINCIPAL DE LA APLICACIÓN
# ==============================================================================

bucle = True
while bucle:

    # Captura la opción del menú, la pasa a mayúsculas y quita espacios accidentales
    operacion = input("¿Qué deseas realizar (escribe la letra correspondiente)? A: visualizar todos los libros de la biblioteca | B: buscar libro | C: \
                    añadir libro | D: eliminar libro por título | E: alquilar libro | F: devolver libro | G: SALIR"
                    ).upper().replace(":", "").strip()

    match operacion:
        
        # ----------------------------------------------------------------------
        # CASO A: VISUALIZAR TODOS LOS LIBROS
        # ----------------------------------------------------------------------
        case "A":
            print(f"NUESTRA BIBLIOTECA TIENE {len(libros)} LIBROS: \n{'-'*100}")
            # Recorremos la lista de libros y mostramos cada par clave-valor con formato limpio
            for libro in libros:
                for clave, valor in libro.items():
                    print(clave, valor, sep=": ")
                print()  # Espacio en blanco entre libro y libro

        # ----------------------------------------------------------------------
        # CASO B: BUSCAR UN LIBRO
        # ----------------------------------------------------------------------
        case "B":
            sigue = True
            while sigue:
                titulo = input("Introduce el título del libro sobre el que deseas información: ").lower().strip()

                # Invocamos la función de búsqueda
                info_libro = busqueda_libro(libros, titulo)

                if info_libro is None:
                    # Si no existe, el bucle 'while sigue' continúa para que lo intente de nuevo
                    print(f"\nNo tenemos el libro titulado '{titulo.title()}' que estás buscando. Inténtalo de nuevo.\n")    
                
                else:
                    sigue = False  # Encontrado, cambiamos el interruptor para salir de este submenú
                    print("\n¡Tenemos el libro que estás buscando! Aquí está su información:")
                    for clave, valor in info_libro.items():
                        print(clave, valor, sep=": ")
        
        # ----------------------------------------------------------------------
        # CASO C: AÑADIR UN NUEVO LIBRO
        # ----------------------------------------------------------------------
        case "C":
            claves = ["Titulo", "Autor", "Alquilado"]
            valores = []
            adicion = {} # Diccionario temporal para armar el nuevo libro
            
            # Forzamos formato Título (Mayúscula en la primera letra de cada palabra)
            titulo = input("Introduce el título del libro que quieres añadir: ").title().strip()
            autor = input("Introduce el/la autor/a del libro: ").title().strip()

            valores.append(titulo)
            valores.append(autor)
            valores.append(False)

            # Mapeamos las dos listas (claves y valores) para construir el diccionario usando bucle e índice
            for i, n in enumerate(claves):
                adicion[n] = valores[i]

            pregunta = f'Confirma que quieres añadir el libro titulado "{titulo.title()}" de {autor} a la biblioteca (si/no): '

            # Si la función de confirmación retorna True, impactamos la lista principal
            if confirmacion(pregunta):
                libros.append(adicion)
                print(f'\n"{titulo.title()}" de {autor} añadido con éxito a la biblioteca.')
            else:
                print(f'\nSe ha cancelado la adición de "{titulo.title()}" de {autor} a la biblioteca.')

        # ----------------------------------------------------------------------
        # CASO D: ELIMINAR UN LIBRO
        # ----------------------------------------------------------------------
        case "D":
            sigue = True
            while sigue:
                titulo = input("Título del libro que quieres eliminar: ").lower().strip()

                # Buscamos el libro completo para obtener el objeto exacto de la memoria
                info_libro = busqueda_libro(libros, titulo)

                if info_libro is None:
                    print(f"\nEl libro que estás intentando eliminar ('{titulo.title()}') no está en la biblioteca.")

                else:
                    sigue = False # Detenemos el reintento de entrada
                    pregunta = f'Confirma que quieres eliminar el libro titulado "{titulo.title()}" (si/no): '

                    if confirmacion(pregunta):
                        # Al pasarle el objeto exacto que retornó la función, .remove() lo borra directamente
                        libros.remove(info_libro)
                        print(f'\n"{titulo.title()}" eliminado con éxito de la biblioteca.')
                    else:
                        print(f'\nSe ha cancelado la eliminación de "{titulo.title()}" de la biblioteca.')

        # ----------------------------------------------------------------------
        # CASO E: ALQUILAR UN LIBRO
        # ----------------------------------------------------------------------
        case "E":
            sigue = True
            while sigue:
                titulo = input("Título del libro que quieres alquilar: ").lower().strip()

                info_libro = busqueda_libro(libros, titulo)
                
                if info_libro is None:
                    print("\nEl libro que estás intentando alquilar no está en la biblioteca.")

                else:
                    sigue = False
                    pregunta = f'Confirma que quieres alquilar el libro titulado "{titulo.title()}" (si/no): '

                    if confirmacion(pregunta):
                        # Comprobamos la clave interna de disponibilidad del diccionario
                        if info_libro["Alquilado"] == False:
                            # Modificación POR REFERENCIA: Cambia directamente en la lista 'libros'
                            info_libro["Alquilado"] = True
                            print(f'\nHas alquilado "{titulo.title()}". Tienes 3 días para recogerlo, si no, se cancela el proceso.')

                        else:
                            print(f'\nEl libro "{titulo.title()}" ya está alquilado.')
                    
                    else:
                        print(f'\nSe ha cancelado el alquiler de "{titulo.title()}".')

        # ----------------------------------------------------------------------
        # CASO F: DEVOLVER UN LIBRO
        # ----------------------------------------------------------------------
        case "F":
            sigue = True
            while sigue:
                titulo = input("Título del libro que quieres devolver: ").lower().strip()
                
                info_libro = busqueda_libro(libros, titulo)
                
                if info_libro is None:
                    print("\nEl libro que estás intentando devolver no está en la biblioteca.")

                else:
                    sigue = False
                    pregunta = f'Confirma que quieres devolver el libro titulado "{titulo.title()}" (si/no): '

                    if confirmacion(pregunta):
                        # Verificamos si el estado actual es alquilado
                        if info_libro["Alquilado"] == True:
                            # Modificación POR REFERENCIA: Cambia directamente en la lista 'libros'
                            info_libro["Alquilado"] = False
                            print(f'\nHas devuelto "{titulo.title()}" con éxito.')

                        else:
                            print(f'\nParece ser que el libro "{titulo.title()}" ya está de vuelta en nuestra biblioteca. Si hay un error,\
                                contacta con nuestro servicio de Atención al Cliente. Gracias.')
                    
                    else:
                        print(f'\nSe ha cancelado el proceso de devolución de "{titulo.title()}".')

        # ----------------------------------------------------------------------
        # CASO G: SALIR DEL SISTEMA
        # ----------------------------------------------------------------------
        case "G":
            bucle = False  # Rompe la condición del while principal y cierra el script
            print("\n*** ¡Hasta lueguito! ***")
```
