## Ejercicio 2
Eleva todos los elementos de la lista al cuadrado
> ```ejer_2 = [1,2,3,4,5]```

## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output:

```['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']```

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]
ejer_3_lista= []

for i in ejer_3_1:
    for j in ejer_3_2:
        x = i + " " + j
        ejer_3_lista.append(x)

print(ejer_3_lista)
````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
ejer_5_vacia = []

for n in ejer_5:
    if n != 3:
        ejer_5_vacia.append(n)

print(ejer_5)
print(ejer_5_vacia)
````

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
copia_ejer_5 = ejer_5.copy()

for i in ejer_5:
    if i == 3:
        copia_ejer_5.pop(copia_ejer_5.index(i))

print(ejer_5)
print(copia_ejer_5)
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]
diccionario_ejer_11 = {}

claves = []
valores = []

for clave, valor in ejer_11:
    #print(clave, valor)
    if clave not in diccionario_ejer_11:
        diccionario_ejer_11[clave] = []
    diccionario_ejer_11[clave].append(valor)
    

print(diccionario_ejer_11)
````

## Ejercicio 15
