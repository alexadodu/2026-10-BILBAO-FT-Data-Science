### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

### Step 3. Name of the max valued product

```python
chipo.info()
```

```python
chipo["item_price"] = chipo["item_price"].str.replace("$","")
chipo["item_price"]= chipo["item_price"].astype(float)

chipo["item_price"]
```

```python
print("Artículo de mayor valor:", chipo["item_name"].loc[chipo["item_price"].idxmax()], chipo["item_price"].max(), "$")
```

```python
chipo.loc[chipo["item_price"].idxmax()]
```

### Step 4. How many products cost more than $10.00?

```python
chipo[chipo["item_price"] > 10].shape[0]
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
# No es lo mismo, porque en un mismo pedido se han comprado más productos y hay que mirar el order_id
chipo[chipo["item_price"] > 10]["order_id"].nunique()
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
# No, hay que sumar todos los productos de un pedido y extraer aquellos que son > 10
pedidos_precio = chipo.groupby("order_id")["item_price"].sum()

print(pedidos_precio.min()) # todos son de más de 10

pedidos_precio[pedidos_precio > 10].size # da el número de pedidos de más de 10, que son todos
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
# No, son productos de más de 10$ que se haya comprado más de 2 veces en un mismo order_id
productos_mas_de_10 = chipo[chipo["quantity"] >= 2]
productos_mas_de_10 = productos_mas_de_10[productos_mas_de_10["item_price"] > 10]
productos_mas_de_10
```

```python
num_pedidos = productos_mas_de_10["order_id"].nunique()

print("El número de pedidos donde se ha comprado un mismo producto de más de 10$ es de", 
      num_pedidos)
```

### Step 5. What is the price of each item and name it unit_price. Get only item_name and unit_price

```python
productos_mas_de_10["unit_price"] = productos_mas_de_10["item_price"]/productos_mas_de_10["quantity"]
extraccion= productos_mas_de_10[["item_name", "unit_price"]]
extraccion
```

### Step 6. Sort by the name of the item

```python
extraccion.sort_values(by="item_name")
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
# VERSION 1
productos_mas_de_10_sorted = productos_mas_de_10.sort_values(by="unit_price", ascending= False).reset_index()
print(f"La candidad del producto más caro ({productos_mas_de_10_sorted["item_name"].loc[0]}: {productos_mas_de_10_sorted["unit_price"].loc[0]}$) ha sido de {productos_mas_de_10_sorted["quantity"].loc[0]}.")
```

```python
# VERSION 2
productos_mas_de_10 = productos_mas_de_10.reset_index()
productos_mas_de_10
```

```python
maximo = productos_mas_de_10["unit_price"].max()
maximo_idx = productos_mas_de_10["unit_price"].idxmax()
# print(maximo)
# print(maximo_idx)

print(f"La candidad del producto más caro ({productos_mas_de_10["item_name"][maximo_idx]}: {maximo}$) ha sido de {productos_mas_de_10["quantity"][maximo_idx]}.")
```
