# Creación de base de datos

```python
import pandas as pd
import sqlite3
```

```python
import os
os.getcwd()
```

```python
# Conectamos con la base de datos cervezas.db
connection = sqlite3.connect("cervezas.db")

# Obtenemos un cursor que utilizaremos para hacer las queries
crsr = connection.cursor()

# Activamos las llaves foráneas, estas son necesarias para que se cumplan las relaciones entre tablas, 
# y que no se puedan insertar datos en una tabla que no existan en otra tabla relacionada
crsr.execute("PRAGMA foreign_keys = ON;")
```

```python
query = '''
 CREATE TABLE IF NOT EXISTS cervezas (
    CodC TEXT PRIMARY KEY,
    Envase TEXT,
    Capacidad FLOAT,
    Stock INTEGER
 );
 '''

crsr.execute(query)
```

```python
query = '''
 CREATE TABLE IF NOT EXISTS bares (
    CodB TEXT PRIMARY KEY,
    Cif TEXT,
    Nombre TEXT NOT NULL,
    Localidad TEXT NOT NULL
 );
 '''

crsr.execute(query)
```

```python
query = '''
 CREATE TABLE IF NOT EXISTS empleados (
    CodE TEXT PRIMARY KEY,
    Nombre TEXT NOT NULL,
    Sueldo INTEGER NOT NULL
 );
 '''

crsr.execute(query)
```

```python
query = '''
 CREATE TABLE IF NOT EXISTS reparto (
    CodE TEXT,
    CodB TEXT,
    CodC TEXT,
    Fecha TEXT NOT NULL,
    Cantidad INTEGER NOT NULL,

    FOREIGN KEY (CodE) REFERENCES empleados(CodE),
    FOREIGN KEY (CodB) REFERENCES bares(CodB),
    FOREIGN KEY (CodC) REFERENCES cervezas(CodC)
 );
 '''

crsr.execute(query)
```

# Ejercicios

1. Obtener el nombre de los empleados que hayan repartido al bar Stop durante la semana del 17 al 23 de octubre de 2005.

```python
query = '''
SELECT DISTINCT empleados.Nombre
FROM empleados
JOIN reparto ON empleados.CodE = reparto.CodE
JOIN bares ON bares.CodB = reparto.CodB
WHERE bares.Nombre = "Stop" AND reparto.fecha BETWEEN "2005-10-17" AND "2005-10-23";
'''
# crsr.execute(query).fetchall()

pd.read_sql_query(query,connection)
```

2. Obtener el Cif y nombre de los bares a los que se ha repartido cerveza de tipo Botella y capacidad inferior a 1 litro, ordenados por localidad.

```python
query = '''
SELECT bares.Cif, bares.Nombre
FROM bares
JOIN reparto ON bares.CodB = reparto.CodB
JOIN cervezas ON cervezas.CodC = reparto.CodC
WHERE cervezas.Envase = "Botella" and cervezas.Capacidad < 1
ORDER BY bares.Localidad;
'''
# crsr.execute(query).fetchall()

pd.read_sql_query(query,connection)
```

3. Obtener los repartos (nombre del bar, envase y capacidad de la bebida, fecha y cantidad) realizados por Prudencio Caminero.

```python
query = '''
SELECT bares.Nombre AS "Bar", cervezas.Envase, cervezas.Capacidad, reparto.Fecha, reparto.Cantidad
FROM reparto
JOIN empleados ON reparto.CodE = empleados.CodE
JOIN cervezas ON cervezas.CodC = reparto.CodC
JOIN bares ON bares.CodB = reparto.CodB
WHERE empleados.Nombre = "Prudencio Caminero";
'''

pd.read_sql_query(query,connection)
```

4. Obtener los bares a los que se les ha repartido envases de tipo botella y capacidad 0.2 ó 0.33.

```python
query = '''
SELECT DISTINCT bares.Nombre AS "Bar"
FROM reparto
JOIN cervezas ON cervezas.CodC = reparto.CodC
JOIN bares ON bares.CodB = reparto.CodB
WHERE cervezas.Envase = "Botella" AND (cervezas.Capacidad = 0.2 OR cervezas.Capacidad = 0.33);
'''

pd.read_sql_query(query,connection)
```

5. Nombre de los empleados que han repartido a los bares "Stop" y "Las Vegas" cervezas con envase botella.

```python
query = '''
SELECT DISTINCT empleados.Nombre
FROM reparto
JOIN empleados ON reparto.CodE = empleados.CodE
JOIN cervezas ON cervezas.CodC = reparto.CodC
JOIN bares ON bares.CodB = reparto.CodB
WHERE (bares.Nombre = "Stop" OR bares.Nombre = "Las Vegas") AND cervezas.Envase = "Botella";
'''

pd.read_sql_query(query,connection)
```

6. Obtener el nombre y número de viajes que ha realizado cada empleado fuera de Villa Botijo.

```python
query = '''
SELECT DISTINCT empleados.Nombre AS "Nombre empleado", COUNT(reparto.Fecha) AS "Número de viajes"
FROM empleados
JOIN reparto ON reparto.CodE = empleados.CodE
JOIN bares ON bares.CodB = reparto.CodB
WHERE bares.Localidad <> "Villa Botijo"
GROUP BY reparto.CodE;
'''

pd.read_sql_query(query,connection)
```

7. Obtener el nombre y localidad del bar que más litros de cerveza ha comprado.

```python
query = '''
SELECT DISTINCT bares.Nombre AS "Nombre bar", bares.Localidad, SUM(reparto.Cantidad * cervezas.Capacidad) AS "Total litros cervezas"
FROM reparto
JOIN bares ON bares.CodB = reparto.CodB
JOIN cervezas ON cervezas.CodC = reparto.CodC
GROUP BY bares.Nombre
ORDER BY "Total litros cervezas" DESC
LIMIT 1;
'''

pd.read_sql_query(query,connection)
```

8. Obtener los bares que han adquirido todos los tipos de cerveza con envase de botella y capacidad menor que 1 litro.

```python
query = '''
SELECT DISTINCT bares.Nombre AS "Nombre bar"
FROM reparto
JOIN bares ON bares.CodB = reparto.CodB
JOIN cervezas ON cervezas.CodC = reparto.CodC
WHERE cervezas.Envase = "Botella" AND cervezas.Capacidad < 1;
'''

pd.read_sql_query(query,connection)
```

9. Subir un 5% el sueldo del empleado que más días haya trabajado.

```python
query = '''
SELECT DISTINCT empleados.Nombre AS "Nombre empleado", COUNT(DISTINCT reparto.Fecha) AS "Días trabajados", empleados.Sueldo
FROM reparto
JOIN empleados ON reparto.CodE = empleados.CodE
GROUP BY empleados.Nombre
ORDER BY "Nombre empleado" DESC;
'''

pd.read_sql_query(query,connection)
```

```python
query = '''
UPDATE empleados
SET Sueldo = Sueldo * 1.05
WHERE CodE = (
    SELECT reparto.CodE
    FROM reparto
    GROUP BY reparto.CodE
    ORDER BY COUNT(DISTINCT reparto.Fecha) DESC
    LIMIT 1
);
'''

crsr.execute(query)
```

```python
query = '''
SELECT DISTINCT empleados.Nombre AS "Nombre empleado", COUNT(DISTINCT reparto.Fecha) AS "Días trabajados", empleados.Sueldo
FROM reparto
JOIN empleados ON reparto.CodE = empleados.CodE
GROUP BY empleados.Nombre
ORDER BY "Nombre empleado" DESC;
'''

pd.read_sql_query(query,connection)
```
