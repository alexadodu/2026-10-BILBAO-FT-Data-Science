```python
import pandas as pd

# 1. Definimos la ruta correcta (con 'r')
ruta = r"C:\Users\alexa\Desktop\Denetik\CURSOS\Data-Science-BBKbootcamp\Carpeta-trabajo\Proyectos\Proyectos_individuales\2-EDA\datos\en.openfoodfacts.org.products.csv.gz"

# 2. Opcional: Define solo las columnas que vas a usar en tu EDA para ir aún más rápido
# (Si quieres TODAS las columnas, borra el parámetro 'usecols' de abajo)
columnas_interes = ['code', 'product_name', 'brands', 'countries_en']

# 3. Inicializamos una lista vacía para ir acumulando los trozos de España
lista_chunks_espana = []

print("Iniciando la lectura por bloques... Esto puede tardar un par de minutos.")

# 4. Leemos el archivo en bloques de 50,000 filas
bloques = pd.read_csv(ruta, sep='\t', chunksize=50000, usecols=columnas_interes, low_memory=False)

for i, chunk in enumerate(bloques):
    # Filtramos el bloque actual: buscamos filas donde 'countries_en' contenga 'Spain'
    # Usamos na=False para evitar errores si hay celdas vacías
    chunk_filtrado = chunk[chunk['countries_en'].str.contains('Spain', case=False, na=False)]
    
    # Si el bloque contiene productos de España, lo añadimos a la lista
    if not chunk_filtrado.empty:
        lista_chunks_espana.append(chunk_filtrado)
    
    # Imprime un contador para saber que el código sigue vivo y avanzando
    if i % 10 == 0:
        print(f"Bloques procesados: {i}...")

# 5. Concatenamos todos los trozos filtrados en un único DataFrame final
df_espana = pd.concat(lista_chunks_espana, ignore_index=True)

print(f"¡Listo! Se ha generado tu DataFrame de España con {df_espana.shape[0]} filas.")
```
