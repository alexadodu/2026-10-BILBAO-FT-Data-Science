````python
print(my_array)
impares_bools = (my_array % 2 != 0)
print(impares_bools)

impares = my_array[impares_bools]
print("\nImpares de my_array (v1):", impares)
````

## Ejercicio 5
En un nuevo array, sustituye todos los impares de `my_array` por -1

````python
my_array_copy = my_array.copy()

my_array_copy[my_array_copy % 2 == 1] = -1

print(my_array_copy)
````

## Ejercicio 6
En un nuevo array, sustituye todos los impares de `my_array` por -1. Esta vez usa `where`

````python
my_array_copy2 = my_array.copy()

my_array_copy2 = np.where(my_array_copy2 % 2 == 1, -1, my_array_copy2)

print(my_array_copy2)
````

## Ejercicio 20
Dado el siguiente array

```Python
x = np.random.randint(10, 20, size = (5,2,6))
```

1. ¿Cuántos elementos tiene el array?
2. ¿Cuántas dimensiones?
3. ¿Cuánta es la longitud cada dimensión?

````python
x = np.random.randint(10, 20, size = (5,2,6))

print("dimensiones: ", x.ndim)
print("nº elementos: ", x.size)

print("longitud de dimensión 1 (nº matrices): ", len(x))
print("longitud de dimensión 2 (nº filas):    ", len(x[0]))
print("longitud de dimensión 3 (nº columnas): ", len(x[0][0]))
````
